```python
import random
import numpy as np
```

```python
%timeit [random.random() for _ in range(100_000)]
```

```python
python_numbers = [random.random() for _ in range(100000)]
numpy_numbers = np.random.random(100000)

print(type(python_numbers))
print(type(numpy_numbers))

print(len(python_numbers))
print(len(numpy_numbers))
```
