# Data Analysis with Pandas and Matplotlib

## Objective
This notebook demonstrates how to:
- Load CSV data using Pandas.
- Inspect and explore a dataset.
- Display data using DataFrames.
- Visualize data using Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
print("Current directory:", Path.cwd())

csv_path = Path("projects/csv-sql-question-workbench/solution/sample.csv")

print("File exists:", csv_path.exists())
print("Resolved path:", csv_path.resolve())
```

```python
df = pd.read_csv(csv_path)

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
df
```

```python
df.describe()
```

```python
product_units = df.groupby("product")["units"].sum()

print(product_units)
```

```python
plt.figure(figsize=(8, 5))

product_units.plot(kind="bar")

plt.title("Total Units by Product")
plt.xlabel("Product")
plt.ylabel("Total Units")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
